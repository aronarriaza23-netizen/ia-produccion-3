```python
from pathlib import Path
import pandas as pd

RAIZ = Path.cwd().parents[0]

df_plan_de_compras = pd.read_excel(RAIZ / "plan_de_compras_2025.xlsx" , sheet_name="PLAN DE COMPRAS 2025")

df_plan_de_compras.head()
```

```python
tabla_tipo_proyecto = (
    df_plan_de_compras
    .groupby("Tipo Proyecto")
    .agg(
        Cantidad_Registros=("ID Proyecto", "count"),
        Cantidad_Proyectos=("ID Proyecto", "nunique"),
        Monto_Total_2025=("Monto Total Ítem Año 2025", "sum"),
        Monto_Promedio_2025=("Monto Total Ítem Año 2025", "mean")
    )
    .reset_index()
)

tabla_tipo_proyecto
```

```python
tabla_unidad_compra = (
    df_plan_de_compras
    .groupby("Unidad de Compra")
    .agg(
        Cantidad_Proyectos=("ID Proyecto", "nunique"),
        Cantidad_Items=("Nombre ítem", "count"),
        Monto_Total_2025=("Monto Total Ítem Año 2025", "sum")
    )
    .reset_index()
    .sort_values(
        "Monto_Total_2025",
        ascending=False
    )
)

tabla_unidad_compra
```

```python
tabla_estado = (
    df_plan_de_compras
    .groupby("Estado Proyecto")
    .agg(
        Cantidad_Registros=("ID Proyecto", "count"),
        Cantidad_Proyectos=("ID Proyecto", "nunique"),
        Monto_Total_2025=("Monto Total Ítem Año 2025", "sum"),
        Monto_Arrastre=("Monto De Arrastre", "sum")
    )
    .reset_index()
)

tabla_estado
```
