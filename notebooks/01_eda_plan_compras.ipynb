# Análisis Exploratorio de Datos - Plan de Compras 2025

El objetivo de este notebook es realizar un análisis exploratorio del Plan de Compras 2025,
con el propósito de conocer la estructura del conjunto de datos, identificar valores
faltantes, registros duplicados, tipos de variables, distribuciones y principales
características de la información.

```python
nulos = pd.DataFrame({
    "Cantidad Nulos": df_plan_de_compras.isnull().sum(),
    "Porcentaje Nulos": (
        df_plan_de_compras.isnull().mean() * 100
    ).round(2)
})

nulos = nulos.sort_values(
    "Porcentaje Nulos",
    ascending=False
)

nulos
```

```python
cantidad_duplicados = df_plan_de_compras.duplicated().sum()

print(f"Registros duplicados: {cantidad_duplicados}")
```

```python
valores_unicos = df_plan_de_compras.nunique().sort_values()

valores_unicos
```

```python
columnas_montos = [
    "Monto Unitario Ítem",
    "Monto Total Ítem Año 2025",
    "Monto De Arrastre"
]

df_plan_de_compras[columnas_montos].describe()
```

```python
df_plan_de_compras[columnas_montos].agg([
    "count",
    "sum",
    "mean",
    "median",
    "min",
    "max"
])
```

```python
for columna in columnas_montos:
    cantidad_negativos = (df_plan_de_compras[columna] < 0).sum()
    print(f"{columna}: {cantidad_negativos} valores negativos")
```
