## 3. Calidad de datos

### 3.1 Valores faltantes

```python
nulos = pd.DataFrame({
    "Cantidad Nulos": df_plan_de_compras.isnull().sum(),
    "Porcentaje Nulos": (
        df_plan_de_compras.isnull().mean() * 100
    ).round(2)
})

nulos = nulos.sort_values(
    "Porcentaje Nulos",
    ascending=False
)

nulos
```

### Comentario
La variable `Meses envio OC de Arraste` presenta **648 valores nulos (77,98%)**, mientras que `OC Asociada Item 2025` presenta **123 valores nulos (14,80%)** y `Código presupuestario` solamente **1 valor nulo (0,12%)**.

El alto porcentaje de nulos de `Meses envio OC de Arraste` no debe interpretarse automáticamente como un problema de calidad. Dado que existe la variable `Item De Arraste`, es posible que el dato solo corresponda a registros que efectivamente sean de arrastre. Por eso conviene comprobar si estos nulos son **estructurales** y coherentes con la lógica de negocio.

```python
# Nulos de "Meses envio OC de Arraste" son coherentes con "Item De Arraste"
pd.crosstab(
    df_plan_de_compras["Item De Arraste"],
    df_plan_de_compras["Meses envio OC de Arraste"].isna(),
    rownames=["Item De Arraste"],
    colnames=["Meses de arrastre es nulo"]
)
```

```python
# Revisa coherencia entre cantidad de OC asociadas y existencia del identificador de OC
tabla_oc_nulos = pd.crosstab(
    df_plan_de_compras["Cantidad OC Asociadas Ítem 2025"] > 0,
    df_plan_de_compras["OC Asociada Item 2025"].isna(),
    rownames=["Tiene OC asociadas según cantidad"],
    colnames=["OC Asociada Item 2025 es nulo"]
)

tabla_oc_nulos
```

Esta tabla permite detectar casos donde se informa una cantidad positiva de órdenes de compra asociadas, pero el identificador de la orden se encuentra vacío, o viceversa.

### 3.2 Registros duplicados

```python
cantidad_duplicados = df_plan_de_compras.duplicated().sum()

print(f"Registros duplicados: {cantidad_duplicados}")
```

```python
# Posibles duplicados desde la lógica de negocio
duplicados_negocio = (
    df_plan_de_compras
    .groupby(["ID Proyecto", "Nombre ítem"], dropna=False)
    .size()
    .reset_index(name="Cantidad registros")
    .query("`Cantidad registros` > 1")
    .sort_values("Cantidad registros", ascending=False)
)

duplicados_negocio.head(20)
```

Los casos anteriores no deben eliminarse automáticamente. Primero se debe comprobar si corresponden a registros legítimos, por ejemplo, distintas órdenes, fechas o condiciones asociadas al mismo ítem.

## 6. Análisis de variables monetarias

```python
columnas_montos = [
    "Monto Unitario Ítem",
    "Monto Total Ítem Año 2025",
    "Monto De Arrastre"
]

df_plan_de_compras[columnas_montos].describe()
```

```python
df_plan_de_compras[columnas_montos].agg([
    "count",
    "sum",
    "mean",
    "median",
    "min",
    "max"
])
```

### Comentario
Las variables monetarias presentan diferencias importantes entre media y mediana. Por ejemplo, `Monto Total Ítem Año 2025` tiene una media cercana a **$101,5 millones** y una mediana de **$35 millones**, mientras que alcanza un máximo de **$4.839 millones**.

Esta diferencia indica una distribución fuertemente asimétrica hacia la derecha: la mayoría de los registros se concentra en montos menores y un grupo reducido de compras de gran magnitud eleva considerablemente el promedio. Por este motivo, la mediana es una medida especialmente útil para representar el monto típico.

```python
for columna in columnas_montos:
    cantidad_negativos = (df_plan_de_compras[columna] < 0).sum()
    print(f"{columna}: {cantidad_negativos} valores negativos")
```

No se observan montos negativos en las tres variables monetarias analizadas. Esto elimina una inconsistencia básica, aunque todavía conviene estudiar montos iguales a cero y valores extremadamente altos.

```python
# Cantidad de valores iguales a cero en las variables monetarias
for columna in columnas_montos:
    cantidad_ceros = (df_plan_de_compras[columna] == 0).sum()
    porcentaje_ceros = (df_plan_de_compras[columna] == 0).mean() * 100
    print(f"{columna}: {cantidad_ceros} ceros ({porcentaje_ceros:.2f}%)")
```
